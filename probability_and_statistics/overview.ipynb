# Probability & Statistics

## Overview

The real world is often too complex and computationally difficult to model or measure with analytical determanism leading to high degrees of uncertainty. Probability allows us to quantify ignorance (how much we dont know), model empirical data in a way that accounts for noise and then make predictions despite our uncertainty.

It is interesting to note that most phenomena we model using probability are actually fundamentally deterministic. It is theoretically possibly for example to predict with 100% certainty the outcome of a coin flip assuming the force of the toss, the air currents in the room, the mass of the coin, etc.. could all be controlled perfectly. For practical purposes however, it is easy to see why such circumstances are virtually impossible to achieve, necessitating probabilistic modelling. 

## Definitions

**Probability:** Assuming a known probability distribution, we attempt to predict specific data points or samples.

**Statistics:** Given a dataset of samples, we attempt to derive the appropriate probabilistic model.

## Examples

- **The behavior of gasses**, impossibly complex interactions of countless particles can be modeled using a handful of statistical quantities like temperature, entropy, etc.. (the physics of thermodynamics)
- **Turbulence** is another example of random, unpredictable and uncertain fluctuations in air currents comprised of multiple variables like countless particles, temperature variations, altitude, etc.. which makes this phenomena a great candidate for statistical and probabilistic modelling
- **Measurement Error.** During an experiment, multiple measurements always have slight variations due to things like human error and instrument imperfections. The error distribution tends to follow the gaussian model.
- **Human Behavior** is highly complex with high degrees of uncertainty.
- **Weather** like turbulence is not deterministic due to computational infeasibility and so is modelled probabilistically.
