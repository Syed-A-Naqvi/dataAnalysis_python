# What is the probability of an event?

The probability of some event $A$ is defined as
$$
    P(A) = \frac{\text{\# of ways $A$ can happen}}{\text{\# of possible things that can happen}}.
$$

# Example: I flip a coin twice.

This is a fair coin where each flip is independent.

```python
Omega = {'HH', 'TT', 'HT', 'TH'} # set of all possible outcomes
```

## 1. A = at least one heads

```python
A = {'HT', 'TH', 'HH'}

print(f"P(A) = {len(A)/len(Omega)}")
```

## 2. B = no heads

```python
B = {'TT'}

print(f"P(B) = {len(B)/len(Omega)}")
```

# Example: I roll two dice.

These are two fair die where each roll is independent.

```python
# Sample space for rolling two dice: all ordered pairs (d1, d2) with d1,d2 ∈ {1,2,3,4,5,6}
Omega = {(i, j) for i in range(1, 7) for j in range(1, 7)}
```

## 1. What is the probability that at least one dice rolls a 5?

```python
A = {(i, 5) for i in range(1,7)}.union({(5, i) for i in range(1,7)})

print(f"P(A) = {len(A)/len(Omega)}")
print(len(A))
```

We could also say
$$
    P(A) = P(D_1 = 5) + P(D_1 \neq 5) \cdot P(D_2 = 5).
$$

## Notes

There are truly random processes such as radioactive decay, and then there are processes such as a coin flips which are theoretically deterministic but the computation required would be infeasible rendering them effectively random.

It is for such systems, weather forecasting, turbulence calculations, stock market predictions, etc... that we use *probability* to model the different outcomes.

For the coin flip example, assuming a fair coin we can calculate apriori using our probability model of 50% chance heads/tails that out of $n$ total flips, there will be roughly $\frac{n}{2}$ heads and $\frac{n}{2}$ tails.

Conversely, we could be presented with a dataset of coin flip outcomes and then use *statistics* to posteriori *hypothesis test* if the coin used to generate the dataset was fair or not. If the outcomes do not support a probability of 50% heads/tails, we can be confident that our initial probability model was wrong and the coin must not be fair.
