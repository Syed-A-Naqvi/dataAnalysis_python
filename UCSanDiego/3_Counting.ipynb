# 3.1 Counting

The number of elements in a set is called its **cardinality** denoted $|\{0,1,2\}| = 3$.

For an inclusive interval where $m \leq n$, $|\{m, \dots, n\}| = m - n + 1$

For a set of integer multiples of $d \in \mathbb{Z}$, defined ${}_d[n] = \{ 1 \leq i \leq n: d | i\}$, the cardinality can be calculated $|{}_d[n]| = \lfloor n/d \rfloor$.

# 3.2 Disjoint Unions

The cardinality of a **disjoint union** is the **sum** of the cardinalities of the individual sets.

- $A \cap B = \emptyset \rightarrow |A \cup B| = | A | + | B |$

- A result that follows: $|\Omega| = |A \cup A^c| = | A | + | A^c | \rightarrow | A | = | \Omega | - | A^c |$ or $| A^c | = | \Omega | - | A |$ known as the **Complement Rule**.

    - If $A$ or $A^c$ is difficult to count, we can use  the **Complement Rule** to calculate the count so long as the complement and universe are known/easy to count.

For example:

- Suppose $A = \{i \in [100] : 3 \nmid i\}$, which implies $A^c = {}_3[100] = \{i \in [100] : 3 \mid i\}$. Since $A$ is difficult to count but we know $|\Omega| = 100$ and $|A^c| = \lfloor 100/3 \rfloor = 33$ we can calculate $|A| = |\Omega| - |A^c| = 100 - 33 = 67$.

The **General Subtraction Rule** says that $A \subseteq B \rightarrow B = A \cup (B - A) \land A \cap (B - A) = \emptyset \rightarrow |B-A| = |B| - |A|$.
