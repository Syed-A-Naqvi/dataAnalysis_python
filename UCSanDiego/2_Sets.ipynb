# 2.1 Notation

A **set** is a collection of elements depicted using **curly braces** where there is **no repetition** and **order doesn't matter**.

- Explicit Set: $\{ \text{Heads}, \text{Tails} \}$, $\{ 0,1 \}$, $\{ a, b, c \}$

- Implicit Set: $\{0, 1, \dots, 9\}$, $\{a, b, \dots, z\}$, $\{\text{Monday}, \dots, \text{Sunday}\}$

- Descriptive Set: $\{ \text{four-letter-words} \}, \{x \in \mathbb{R} \, | \, x^{2} < 4\}$

An **ordered tuple** is a collection of **ordered elements**:

- (1, 0) $\neq$ (0, 1)

# 2.2 Basic Sets

Some familiar sets:

- $\mathbb{Z}$ set of integers 
- $\mathbb{N}$ set of natural numbers
- $\mathbb{E}$ set of even numbers
- $\mathbb{P}$ set of positive numbers
- $\mathbb{Q}$ set of rational numbers
- $\mathbb{R}$ set of real numbers

**Divisibility** says that for $m, n \in \mathbb{Z}$, if $n$ can be expressed as $n = c \cdot m$ for some $c \in \mathbb{Z}$ then we say $n$ is a multiple of $m$ or that $m$ divides $n$ and we write $m \mid n$.

- $6 = 2 \cdot 3 \rightarrow 3 \mid 6$ or $6$ is a multiple of $3$.

- $-8 = (-2) \cdot 4 \rightarrow 4 \mid -8$ or $-8$ is a multiple of $4$.

- $0 = 0 \cdot (-2) \rightarrow -2 \mid 0$ or $0$ is a multiple of $-2$.

If no such $c \in \mathbb{Z}$ exists, we say $m$ does not divide $n$ or $n$ is not a multiple of $m$ denoted $m \nmid n$.

- $\not\exists \, c \in \mathbb{Z}$ such that $4 = c \cdot 3 \rightarrow 3 \nmid 4$.

- $ 0 \nmid n$ for any $n \neq 0$.

**Sets of Multiples** are defined for $m \in \mathbb{Z}$ such that ${}_m\mathbb{Z} \overset{\text{def}}{=} \{i \in \mathbb{Z} : m \mid i\}$:

- ${}_2\mathbb{Z} = \{ \dots, -4, -2, 0, 2, 4, \dots\} = \mathbb{E}$ (even numbers)

- ${}_1\mathbb{Z} = \{ \dots, -2, -1, 0, 1, 2, \dots\} = \mathbb{Z}$ (integers)

- ${}_0\mathbb{Z} = \{0\}$

for $n \in \mathbb{P}$, we have ${}_m[n] = \{i \in \{1, \dots, n\} : m \mid i \}$

- ${}_3[13] = \{i \in \{1, \dots, 13\} : m \mid i \} = \{3, 6, 9, 12\}$

# 2.4 Relations

Two sets are **equal**, denoted $A = B$ if they have exactly the same number of elements:
- for example {0,1} = {1,0}

```python
A = {0,1}
B = {1,0}

venn.venn2((A,B), set_labels=('A', 'B'))
plt.show()
print(A == B)
```

Two sets are **not equal** $A \neq B$ iff $\exists x : (x \in A \land x \notin B) \lor (x \in B \land x \notin A)$

```python
A = {0,1}
B = {1,0,2}

venn.venn2((A,B), set_labels=('A', 'B'))
plt.show()
print(A == B)
```

Two sets **intersect** iff they share at least one common element $\rightarrow A \cap B \neq \emptyset \iff \exists x: \; x \in A \land x \in B$.

```python
A = {0,1,4}
B = {0,2,3,6,7}

venn.venn2((A,B), set_labels=('A', 'B'))
plt.show()
print(A.intersection(B))
```

Two sets are **disjoint** iff they share no elements $\rightarrow A \cap B = \emptyset \iff \neg \exists x: x \in A \land x \in B$.

```python
A = {0,1,2,3}
B = {4,5}

venn.venn2((A,B), set_labels=('A', 'B'))
plt.show()

print(f'The intersection is empty -> {A.intersection(B)}')
```

A **non-empty** universe $\Omega$ intersects every set, $\emptyset$ is **disjoint** from all sets and a set intersects **itself** iff its non-empty.

We can also generalize, deducing that several sets intersect iff they **all** share a common element and are **mutually disjoint** if every pair combination is disjoint.

---

The notion of the **subset** is a generalization of <=.
If every element of $A$ is in $B$ then $A$ is a subset of $B \rightarrow$ ($A \subseteq B$) or $B$ is a superset of $A \rightarrow$ ($B \supseteq A$).
If $A$ has an element not in $B$, $A$ is **not** a subset of $B \rightarrow$ ($A \not\subseteq B$) and $B$ is **not** a superset of $A \rightarrow$ ($B \not\supseteq A$).

```python
B = {0,1,2,3}
A = {1,2}

venn.venn2((A,B), set_labels=('A', 'B'))
print(A.issubset(B))
plt.show()

B = {0,1,2,3}
A = {1,2,5}

venn.venn2((A,B), set_labels=('A', 'B'))
print(A.issubset(B))
plt.show()
```

Note the following results:
- $\mathbb{P} \subseteq \mathbb{N} \subseteq \mathbb{Z} \subseteq \mathbb{Q} \subseteq \mathbb{R}$.

- $\emptyset \subseteq A \subseteq A \subseteq \Omega$. (the empty set is a subset of all sets and all sets are a subset of the universe)

- $A \subseteq B \land B \subseteq C \rightarrow A \subseteq C$. (the subset is **transitive**)

- $A \subseteq B \land B \subseteq A \rightarrow A = B$.

---

The notion of **strict subset** is a generalization of <.

If $A \subseteq B \land A \neq B \rightarrow A \subset B \land B \supset A$.

If $A \not\subset B$ then either $A \not\subseteq B \rightarrow \exists x: x \in A \land x \not\in B$ or $A = B$.

# 2.5 Operations

A **complement** of a set $A$ denoted $A^c$ is defined $\{ x \in \Omega : x \not\in A\}$:
- $\emptyset^c = \Omega$
- $\Omega^c = \emptyset$
- $A^c \cap A = \emptyset$
- $(A^c)^c = A$ (involution)
- $B \subseteq A \rightarrow B^c \supseteq A^c$

The **intersection** $A \cap B$ is the set containing all elements in both $A$ and $B \rightarrow \{ x \in \Omega : x \in A \land x \in B\}$.
- $\bigcap_{i=1}^{t} A_i = \{x: \forall i \in [1,t], x \in A_i\}$

The **union** $A \cup B$ is the set containing all elements in $A$ or $B \rightarrow \{ x \in \Omega : x \in A \lor x \in B\}$.
- $\bigcup_{i=1}^{t} A_i = \{x: \exists i \in [1,t], x \in A_i\}$

The **symmetric difference** $A \Delta B$ is the **XOR** equivalent in set theory containing all elements in either $A$ or $B$ but not both $\rightarrow \{ x \in \Omega : (x \in A \cup B) \land (x \not\in A \cap B)\}$.
- $\Delta_{i=1}^{t} A_i = \{x: (x \in \bigcup_{i=1}^{t} A_i) \land (x \not\in \bigcap_{i=1}^{t} A_i) \}$

- $A \Delta B = (A - B) \cup (B - A)$

**Identity:** $A \cap \Omega = A$ and $A \cup \Omega = \Omega$

**Universal Bound:** $A \cap \emptyset = \emptyset$ and $A \cup \emptyset = A$

**Idempotent:** $A \cap A = A$ and $A \cup A = A$

**Complement:** $A \cap A^c = \emptyset$ and $A \cup A^c = \Omega$

**Commutative:** $A \cap B = B \cap A$ and $A \cup B = B \cup A$

**Associative:** $(A \cap B) \cap C = A \cap (B \cap C)$ and $(A \cup B) \cup C = A \cup (B \cup A)$

**Distributive:** $A \cap (B \cup C) = (A \cap B) \cup (A \cap C)$ and $A \cup (B \cap C) = (A \cup B) \cap (A \cup C)$

**DeMorgan:** $(A \cap B)^c = A^c \cup B^c$

A set **difference** $A-B$ or $A \setminus B$ is the set of elements in $A$ and **not** in $B$ $\rightarrow A-B = A \setminus B = \{ x: x \in A \land x \not\in B\}$.
- Note also that $A - B = A \cap B^c$.

```python
A = {1,2,3}
B = {8, 9}
C = {1, 10, 12}
```

```python
print(f"A u B = {A.union(B)}")
print(f"A n C = {A.intersection(C)}")
print(f"A - C = {A - C}")
```

# 2.6 Cartesian Products

A **cartesian product** of sets $A$ and $B$ is denoted $A \times B$ and is defined as the set $\{ (a,b) : a \in A \land b \in B \}$. It is every combination of tuples where the first element belongs to $A$ and the second element belongs to $B$.
- $A \times A = A^2$

- $\mathbb{R}^2 = \{ (x,y) : x,y \in \mathbb{R} \} \leftarrow $ (cartesian plane)

- $A, B \subseteq \mathbb{R} \rightarrow A \times B \subseteq \mathbb{R}^2$

**Discrete sets** are finite or countably infinite sets with finite or countably infinite cartesian products:

- $ \{a, b\} \times \{ 1, 2, 3\} = \{(x,y) : x \in \{ a, b\} \land y \in \{ 1, 2, 3\} \}$

    

- $A \times \emptyset = \emptyset \times A = \emptyset$

- $A \times (B \cup C) = A \times B \cup A \times C$

    

Cartesian product can be generalized to multiple sets:

- $A \times B$ is a 2-dimensional rectangle

- $A \times B \times C$ is a 3-dimensional cuboid where each square represents an element of a 2-set cartesian product and each block represents an element of a 3-set cartesian product

    

**Sequences** are just tuples without brackets and sometimes without commas.
