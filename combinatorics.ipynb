# Why count?

For many probability calculations based on our simple formula
$$
P(A) = \frac{|A|}{|\Omega|},
$$
the challenge is in counting the total number of ways things can happen which can get very complex.

For example, given a hand of 5 cards from a standard deck, suppose we wish to determine the probability of event $A$ where irrespective of order, we end up with 4 aces and 1 king. First we count how many ways event $A$ can happen, then we divide this value by the *sample space ($\Omega$)* or the total number of 5-card-hands that are possible given a standard deck of cards.

Since we don't care about order, $|A|$ is just the different suits of kings which is 4 while $|\Omega|$ is $52 \times 51 \times 50 \times 49 \times 48$ where there are 52 possibilities for the first card, 51 for the second and so on. 

$$
    \therefore P(A) = \frac{|A|}{|\Omega|} = \frac{4}{52!/(52-5)!}
$$

```python
print(f"P(A) = {4/(math.factorial(52)/math.factorial(52-5))}")
```

## Exercise 1

### \# of possible sequences of 10 coin flips (order matters)

This problem illustrates how quickly it becomes impractical to count outcomes one at a time. Instead we note that each coin has 2 possible outcomes, meaning 10 coins would have $2^{10}$ outcomes.

This is a "with replacement" counting problem since each coin flip is independent and outcomes are not eliminated from the pool of possible outcomes for subsequent flips. 

```python
print(f"Number of 10 flip sequence variations = {2**10}")
```

## Exercise 2

### \# of possible 5 card sequences from a standard 52 card deck (order matters)

52 possibilities for the first slot, 51 for the second, 50 for the third, 49 for the fourth and 48 for the fifth $\rightarrow$
$$
    52\times51\times50\times49\times48 = \frac{52!}{(52-5)!}.
$$

This is a "without replacement" problem since each selected card is removed from the pool of total cards. This makes each card selection a "dependent event". 

```python
print(f"Number of possible 5 card sequence variations = {(math.factorial(52)/math.factorial(52-5))}")
```

## Formulas

We can generailze the two examples above with $n$ choices and $r$ samples into the following useful formulas:

$$
\begin{align}
    \text{With replacement (Order matters):}    \quad & n^r \\
    \text{Without replacement (Order matters):} \quad & \frac{n!}{(n-r)!} = nPr
\end{align}
$$

## Note

Suppose we revisit the 5-card-hand example and change it so that order doesn't matter. We note that any 5-card selection now has $5!$ redundant permutations that are being over-counted in equation $(2)$. To count all $5!$ permutations of any set of 5 cards only once, we simply divide $(2)$ by $5!$.

We generalize as follows:

$$
\begin{align}
    \text{Without replacement (Order doesn't matter):} \quad & \frac{n!}{r!(n-r)!} = {n \choose r} = nCr \tag{3}
\end{align}
$$

## Exercise 3

### How many total possible license plates with 4 letters followed by 2 numbers?

This is essentially the product of two permutation (order matters) with replacement counting problems.

For each license plate there are $r_1=4$ slots with $n_1=26$ choices of letters and $r_2=2$ slots with $n_2=10$ choices of numbers. We calculate the total permutations as follows:

$$
    n_1^{r_1} \times n_2^{r_2} = 26^4 \times 10^2.
$$



```python
print(f"Total unique license plates with repeating numbers = {26**4 * 10**2}")
```

## Exercise 4

We could also consider the case were you are not allowed to repeat letters or numbers. This then becomes the product of two applications of equation $(2)$ above:

$$
    n_1Pr_1 \times n_2Pr_2 = \frac{n_1!}{(n_1-r_1)!} \times \frac{n_2!}{(n_2-r_2)!} = \frac{26!}{(26-4)!} \times \frac{10!}{(10-2)!}
$$

```python
print(f"Total unqiue license plates without repeating numbers = {math.factorial(26)/math.factorial(26-4) * math.factorial(10)/math.factorial(10-2)}")
```

## Exercise 5

### What is the probability of getting a license plate where all letters are unique?

This is a wonderful application of both with and without replacement permutation counting problems tied back to probability.

Note that the *size* of the sample space $(|\Omega|)$ for this problem is actually the result of exercise 3 where we include both repeating and non-repeating letter plates in our count.

If we now let $A$ be the event we find a plate with non-repeating letters, $|A|$ is a product where equation $(2)$ is used for the letters while $(1)$ is used for the numbers since these may still repeat. The probability calculation is as follows:

$$
    P(A) = \frac{|A|}{|\Omega|} = \frac{\frac{26!}{(26-4)!} \times \cancel{10^2}} {26^4 \times \cancel{10^2}} = \frac{\frac{26!}{(26-4)!}} {26^4}
$$

```python
print(f"P(A) = { (math.factorial(26)/math.factorial(26-4)) / (26**4) }")
```
